**Interactive Sentiment Analysis**

```python
from transformers import pipeline
import logging
logging.getLogger("transformers").setLevel(logging.ERROR)
analyzer = pipeline("sentiment-analysis", model="distilbert/distilbert-base-uncased-finetuned-sst-2-english")
print("--- Simple Sentiment Analysis ---")
print("Type 'exit' to stop, or just press the Stop button in Colab.\n")
try:
    while True:
        user_text = input("Enter your text: ")
        if user_text.lower() == 'exit':
            print("Done!")
            break
        if user_text.strip():
            result = analyzer(user_text)[0]
            confidence = round(result['score'] * 100, 1)
            print(f"➜ {result['label']} ({confidence}%)\n")
except KeyboardInterrupt:
    print("\nProgram stopped safely.")
```
